# 12 - Account B: appearance benchmark
Appearance backends (deca, opendelight, trust) x 20 subjects x 5 conditions. Each attempt saves its UV texture, render and metrics; a backend whose code or weights are missing is recorded as `not_run` with the reason. Output goes to `/kaggle/working/out_B` and is zipped.

```python
REPO_URL = "https://github.com/saltypal/TrueFace3D.git"   # <- set before running
import os, subprocess, sys
REPO = "/kaggle/working/TrueFace3D"
if os.path.isdir(REPO + "/.git"):
    subprocess.run(["git", "-C", REPO, "pull", "--ff-only"], check=True)
else:
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO], check=True)
print("commit", subprocess.run(["git", "-C", REPO, "rev-parse", "HEAD"], capture_output=True, text=True).stdout.strip())
os.chdir(REPO); sys.path.insert(0, REPO)
```
